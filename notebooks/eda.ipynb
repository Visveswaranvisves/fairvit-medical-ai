```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/metadata.csv")
df.head()
```

```python
label_counts = df["label"].value_counts()
print(label_counts)

label_counts.plot(kind="bar")
plt.title("Class Distribution (Disease vs No Disease)")
plt.xlabel("Label")
plt.ylabel("Count")
plt.show()
```

```python
group_label_counts = df.groupby("sex")["label"].value_counts().unstack()
print(group_label_counts)

group_label_counts.plot(kind="bar", stacked=True)
plt.title("Disease Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
df["age_group"] = pd.cut(df["age"], bins=[0,40,60,100], labels=["Young","Middle","Old"])
age_group_counts = df["age_group"].value_counts()
print(age_group_counts)

age_group_counts.plot(kind="bar")
plt.title("Age Group Distribution")
plt.show()
```
