```python
from torchvision import transforms

# Image size expected by ViT
IMAGE_SIZE = 224

train_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),  # data augmentation
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])

val_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])
```

```python
import sys
import os

sys.path.append(os.path.abspath(".."))
import pandas as pd
from src.dataset import ChestXrayDataset

train_df = pd.read_csv("../data/processed/train_metadata.csv")
val_df = pd.read_csv("../data/processed/val_metadata.csv")

train_dataset = ChestXrayDataset(
    dataframe=train_df,
    image_dir="../data/raw/images",
    transform=train_transforms
)

val_dataset = ChestXrayDataset(
    dataframe=val_df,
    image_dir="../data/raw/images",
    transform=val_transforms
)

print("Train samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
```

```python
from torch.utils.data import DataLoader

BATCH_SIZE = 16  # safe for most GPUs

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2
)
```

```python
images, labels, sexes = next(iter(train_loader))

print("Image batch shape:", images.shape)
print("Labels shape:", labels.shape)
print("First 5 sexes:", sexes[:5])
```
