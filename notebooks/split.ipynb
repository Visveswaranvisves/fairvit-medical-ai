```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load metadata
df = pd.read_csv("../data/processed/metadata.csv")

# Create a combined stratification column
df["stratify_col"] = df["label"].astype(str) + "_" + df["sex"]

# Train-validation split (80/20)
train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["stratify_col"]
)

print("Train size:", len(train_df))
print("Validation size:", len(val_df))
```

```python
print("Train distribution:")
print(train_df["stratify_col"].value_counts())

print("\nValidation distribution:")
print(val_df["stratify_col"].value_counts())
```

```python
train_df.to_csv("../data/processed/train_metadata.csv", index=False)
val_df.to_csv("../data/processed/val_metadata.csv", index=False)
```
